## Arrays can be of n dimensions 

```python
my_matrix=[100,200,300,400,500]
print(my_matrix)
matrix=np.array(my_matrix)
print(matrix)
print("dimensions of matrix:",matrix.ndim)
print("type of the array:",type(matrix))
print("shape of the array:",matrix.shape)
print("size of the array :",matrix.size)
b=matrix.tolist()
print(type(b))
```
