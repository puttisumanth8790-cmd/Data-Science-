```python
def is_palindrome(s): # is palindrome using for loop 
    reversed_string=""
    for char in s:
        reversed_string=char+reversed_string
    if s==reversed_string:
        return True
    else:
        return False
print(is_palindrome("madam"))
```

```python
# kmeans implementation 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import warnings 
warnings.filterwarnings("ignore")
```

```python
data={
    "Person":["person1","person2","person3","person4","person5","person6"],
    "QSAL":[20,135,75,23,70,140],
    "QSA2":[25,150,50,20,42,135]
}
df=pd.DataFrame(data)
print(df)
```

```python
plt.scatter(df["QSAL"],df["QSA2"])
plt.title("Quatarly sales")
plt.xlabel("QASL")
plt.ylabel("QSA2")
plt.show()
```

```python
df
```

```python
df=df[["QSAL","QSA2"]]
df
```

```python
kmeans=KMeans(n_clusters=3,random_state=42)
kmeans.fit(df)
```

```python
df["cluster"]=kmeans.labels_
print(df)
```

```python
print("centroids")
print(kmeans.cluster_centers_)
```

```python
plt.scatter(df["QSAL"],df["QSA2"],
            c=df["cluster"],
            cmap="viridis",
            s=100)
plt.scatter(kmeans.cluster_centers_[:,0],
            kmeans.cluster_centers_[:,1],
            color="red",
            marker="x",
            s=250)
plt.xlabel("QSAL")
plt.ylabel("QSA2")
plt.title("k-means clustring")
plt.show()
```
