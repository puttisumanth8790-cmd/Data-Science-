```python
# write a function to reverse a string in python 
def reverse_string(s): # def is used to create a function in python and s is the parameter
    return s[::-1] # start stop and step slicing technique 
print(reverse_string("hello")) # then passing the input to the function 
```

```python
def h(s):
    return s[::-1]
print(h("sumanth and harishi "))
```

```python
def reversed_string(s):
    reversed_string=""
    for char in s:
        reversed_string=char+reversed_string
    return reversed_string
print(reversed_string("hello"))
```

```python
def sumanth(s): # using for loop 
    sumanth=""
    for a in s:
        sumanth=a+sumanth
    return sumanth
print(sumanth("hello"))
```

```python
def sumanth(s): # python has built in function called reversed 
    return "".join(reversed(s))
print(sumanth("llohe"))
```
