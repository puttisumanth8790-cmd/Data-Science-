```python
def reversed_string(s):
    reversed_string=""
    for char in s:
        reversed_string=char+reversed_string
    return reversed_string
print(reversed_string("hello"))
```

```python
def sumanth(s): # using for loop 
    sumanth=""
    for a in s:
        sumanth=a+sumanth
    return sumanth
print(sumanth("hello"))
```

```python
def is_palindrome(s): # is palindrome using for loop 
    reversed_string=""
    for char in s:
        reversed_string=char+reversed_string
    if s==reversed_string:
        return True
    else:
        return False
print(is_palindrome("madam"))
```

```python
def sumanth(s):
    hari=""
    for i in s:
        hari=i+hari
    if s==hari:
        return True
    else:
        return False
print(sumanth("madam"))
```

```python
def fibonacci(n):
    if n<=1:
        return n 
    else:
        return fibonacci(n-1)+fibonacci(n-2)
n=5
print(f"The {n}th fibonacci number is :{fibonacci(n)}")
```
