What are duplicates?

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
import warnings
warnings.filterwarnings("ignore")
```

```python
data={"Id":[101,102,103,102,104],
      "Name":["Rahul","Priya","Amit","Priya","Neha"],
      "Age":[25,28,30,28,26],
      "City":["Delhi","Mumbai","Pune","Mumbai","Delhi"]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.duplicated()) # a simple command for printing the duplicates 
# if any duplicate is present it shows true either it shows false 
```
