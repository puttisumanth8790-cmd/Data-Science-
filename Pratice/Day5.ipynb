What are duplicates?
1) Aduplicate entry means the same record appers more than once in ypur dataset 

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
import warnings
warnings.filterwarnings("ignore")
```

```python
data={"Id":[101,102,103,102,104],
      "Name":["Rahul","Priya","Amit","Priya","Neha"],
      "Age":[25,28,30,28,26],
      "City":["Delhi","Mumbai","Pune","Mumbai","Delhi"]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.duplicated()) # a simple command for printing the duplicates 
# if any duplicate is present it shows true either it shows false 
```

```python
duplicates=df[df.duplicated()]
print(duplicates) #  The way to showing only duplicates rows 
```

```python
df[df.duplicated()] # used to find the occurance of all duplicates 
```

```python
df[df.duplicated(keep=False)] #  its shows both copies 
# Keep=False "mark Every occurance of a duplicate as tuple"
```

```python
# checking duplicates based on column 
print(df[df.duplicated("Id",keep=False)])
# this is extremely useful for finding duplicate customer ids employee ids ,product ids
#transaction ids,etc
```

```python
# checking duplicates using multiple colomns 
print(df[df.duplicated(["Name","Age","Id"],keep=False)])
# we want to check for thr  multiple colomns we need to give the square brakets  
```

```python
# count the number of duplicate rows 
print(df.duplicated().sum())
```

```python
# remove duplicates 
print("duplicates removed as per ur requirement")
data=df.drop_duplicates()
print(data)
```

```python
# removing duplicates based up on columns
print(df.drop_duplicates(subset=["Name"],keep=False))
```
