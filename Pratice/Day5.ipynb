# what is correlation?
1) correlation tells us how strongly two numerical variables are related to each other
* correlation generally ranges from -1 to +1 
* +1 perfect positive relationship 
* +0.7 to +1 Strong Postive relation 
* +0.3 to +0.7 Moderate positive relationship
* 0 no linear relationship
* -0.3 to 0 weak negative relationship
* -0.7 to -0.3 Moderate negative relationship
* -1 perfect negative relationship 


```python
data1={
    "hourse_studied":[2,3,4,5,6,7,8],
    "Exam_Score":[50,55,60,65,70,75,80]
}
df=pd.DataFrame(data1)
print(df)
```

```python
# method1 using .corr
correlation=df["hourse_studied"].corr(df["Exam_Score"])
print(correlation) # WE are getting the perfect correlation 
```

```python
data = {
    "Exercise_Hours": [1, 2, 3, 4, 5],
    "Weight": [80, 77, 74, 71, 68]
} # this perfect negative correlation 
pf = pd.DataFrame(data)
print(pf["Exercise_Hours"].corr(pf["Weight"]))
```
