## Filtering the data 

```python
import pandas as pd 
```

```python
df=pd.read_csv(r"D:\Downloads\Details.csv")
```

```python
df.head()
```

```python
df[df["Amount"]>4000] # selecting one column
```

```python
df[
    (df["Amount"]>4000)&
    (df["Category"]=="Electronics")
    ] ## selecting multiple conditions 
```

```python
result=df.query("Amount>4000")
```

```python
result
```

```python
results=df.query("Amount>4000 and Quantity>10")
```

```python
results
```

```python
results=df[df["Category"].isin(["Furniture","Electronics"])]
```

```python
results
```

```python
results=df[~df["Category"].isin(["Furniture","Electronics"])]
print(results)
```
