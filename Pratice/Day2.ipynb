## Matplotlib

```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
data={
    "Month":["jan","feb","Mar","Apr","May","jun"],
    "sales":[12000,15000,14000,18000,21000,19000],
    "Profit":[2000,2500,2200,3200,4000,3500],
    "Customer":[120,150,140,180,210,195]
}
df=pd.DataFrame(data)
print(df)
```

```python
plt.plot(df["Month"],df["sales"]) ## line chart
plt.title("monthly Sales")
plt.xlabel("Month")
plt.ylabel("sales")
```

```python
plt.bar(df["Month"],df["sales"]) ## bar chart 
plt.title("Monthly sales")
plt.xlabel("Month")
plt.ylabel('sales')
```

```python
plt.scatter(df["Customer"],df["sales"]) ## scatter plot
plt.title("Customer vs sales")
plt.xlabel("Customer")
plt.ylabel("sales")
```

```python
plt.hist(df["sales"])  ## displaying histogram 
plt.title("sales distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.boxplot(df["sales"]) ## box plot
plt.title("sales Box plot")
plt.ylabel("sales")
plt.show()
```

```python
plt.pie(df["sales"],labels=df["Month"],autopct="%1.1f%%")
plt.title("Sales by product")
plt.show()
```

```python
plt.plot(df["Month"],df["sales"])
plt.title("Monthly sales") ## title
plt.show()
```

```python
df.columns
```

```python
plt.plot(df["Month"],df["sales"],label="sales")  ## legend
plt.plot(df["Month"],df["Profit"],label="Profit")
plt.legend()
plt.show()
```

```python
plt.plot(df["Month"],df["sales"]) ## using grid lines 
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(10,5)) ## figsize
plt.plot(df["Month"],df["sales"])
plt.show()
```

```python
plt.plot(df["Month"],df["sales"])
plt.title("Monthly sales")
plt.savefig("monthly_sales.png")
plt.show() ## saveing images 
```

```python
plt.plot(df["Month"],df["sales"]) 
plt.ylim(0,25000) ## used to control the axis line 
plt.show()
```

```python
fig,ax=plt.subplots() ## subplots 
ax.plot(df["Month"],df["sales"])
ax.set_title("Monthly sales")
ax.set_xlabel("Month")
ax.set_ylabel("sales")
plt.show()
```

```python
fig,ax=plt.subplots(1,2,figsize=(12,5)) ## multiple plots with subplots()
ax[0].plot(df["Month"],df["sales"])
ax[0].set_title("sales")

ax[1].plot(df["Month"],df["Profit"])
ax[1].set_title("profit")
plt.show()
```

```python
plt.barh(df["Month"],df["sales"]) ## horizontal bar chart
plt.title("Monthly sales")
plt.xlabel("sales")
plt.ylabel("Month")
plt.show()
```
