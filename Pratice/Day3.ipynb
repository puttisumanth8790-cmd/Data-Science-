```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
warnings.filterwarnings("ignore")
```

```python
data={
    "Department":["IT","IT","HR","HR","Sales","Sales","IT","Sales"],
    "Gender":["Male","Female","Male","Male","Female","Male","Female","Female"],
    "Salary":[40000,50000,60000,700000,50000,40000,50000,60000],
    "Age":[26,25,27,28,29,30,33,55],
    "Experience":[2,3,4,5,3,2,4,5],
    "Sales":[5000,6000,7000,3000,2000,5000,4000,6000],
}
df=pd.DataFrame(data)
print(df)
```

```python
sns.lineplot(data=df,x="Experience",y="Salary") ## line plot
plt.title("Experience vs salary")
plt.show()
```

```python
sns.barplot(data=df,x="Department",y="Salary")  ## bar plot
plt.title("Average Salary by Department")
plt.show()
```

```python
sns.countplot(data=df,x="Department")  ## countplot
plt.title("Number of employees by Department")
plt.show()
```

```python
sns.scatterplot(data=df,x="Experience",y="Salary") ## scatterplot
plt.title("Experience vs Salary")
plt.show()
```

```python
sns.histplot(data=df,x="Salary") ## histplot
plt.title("salary Distribution")
plt.show()
```

```python
sns.boxplot(data=df,x="Department",y="Salary") ## boxplot
plt.title("salary distribution by department")
plt.show()
```

```python
sns.violinplot(data=df,x="Department",y="Salary") ## violinplot
plt.title("Salary Distribution ")
plt.show()
```

```python
sns.kdeplot(data=df,x="Salary") ## kdeplot
plt.title("Salary Density")
plt.show()
```

```python
sns.regplot(data=df,x="Experience",y="Salary") ## regplot
plt.title("Experience vs Salary") 
plt.show()
```

```python
correlation=df[["Age","Salary","Experience","Sales"]].corr()
print(correlation)
```

```python
sns.heatmap(correlation,annot=True) ## heatmap
plt.title("correlation Matrix")
plt.show()
```

```python
sns.pairplot(df) ## pairplot
plt.show()
```

```python
sns.jointplot(data=df,x="Experience",y="Salary") ## jointplot
plt.title("Department wise Salary")
plt.show()
```

```python
sns.catplot(data=df,x="Department",y="Experience",kind="bar") ## catplot
plt.show()
```

```python
sns.set_theme() ## setting theme ## hue and size
sns.scatterplot(data=df,x="Experience",y="Salary",hue="Gender",size="Sales") 
plt.show()
```
