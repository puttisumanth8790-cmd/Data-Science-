```python
# Numpy 
# pip install numpy 
import numpy as np 
arr1=np.array([])
print(arr1) # displaying the empty array
arr2=[100,200,300,400] # craeting the list 
df=np.array(arr2) # coverting the list in to np array 
print(df)# printing the np array 
# we have to check the type of the array 
print(type(df))
# We are checking the dimension of the array 
print(df.ndim) # it has only one dimension 
# printing the size of the array 
print(df.size) # the size of the array id 4
# checking the shape of the array 
print(df.shape) # the shape 4
# concatenating the list 
arr2+[200] # adding the data to existing list 
# checking the datatype 
print(df.dtype) # the datatype is int64
# we have to check the mean of the array 
print(np.mean(df)) # the mean is 250
# reshapping 
print(df.reshape(2,2)) # we reshape to 2 rows to two columns 
# flatten function it use to convert the n dimenssional array in to one dimensional array 
print(df.flatten()) # its converted 
#Arange concept 
print(np.arange(15))
print(np.arange(2,100,10)) # three are same with small difference we know that difference 
print(np.arange(2,50))
# we want to print one d array with zeros 
print(np.zeros(3))
print(np.zeros((1,10)))
# we can ones with n dimenssional 
print(np.ones((2,5,6)))
# lines space 
print(np.linspace(1,15))
print(np.linspace(2,25,10,retstep=True))
# creatig the identity matrix
print(np.eye(2,2))
# boarding casting 
print(df+6)
# copy 
pf=df.copy() # copy function retains the original copy and creates the backup 
print(pf)
# random numbers 
print(np.random.rand(2,2))
```
