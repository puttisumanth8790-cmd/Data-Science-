```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

df = pd.read_excel(r"D:\Sumanth.xlsx")

# Clean column names and convert numeric fields
df.columns = df.columns.str.strip()
df["Attendance"] = pd.to_numeric(df["Attendance"], errors="coerce")
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

# Drop rows with missing or infinite values before fitting the model
clean_df = df.replace([np.inf, -np.inf], np.nan).dropna(subset=["Attendance", "Marks"])

X = clean_df[["Attendance"]]
y = clean_df["Marks"]

# Add intercept
X = sm.add_constant(X)

model = sm.OLS(y, X)
results = model.fit()

print(results.summary())
```

```python
predictions=results.predict(X)
print(predictions)
```

```python
import statsmodels.api as sm ## logistic Regression
# Convert Marks into a binary outcome for logistic regression.
# Replace the threshold below with the business rule you want to model.
threshold = 50

y_bin = (clean_df["Marks"] >= threshold).astype(int)

model = sm.Logit(y_bin, X)
result = model.fit()

print(result.summary())
```

```python
from scipy.stats import ttest_1samp # Hypothesis testing
marks = [70,75,80,85,90]
t_stat, p_value = ttest_1samp(marks,70)
print(t_stat)
print(p_value)
```

```python
from statsmodels.formula.api import ols ## ANOVA
from statsmodels.stats.anova import anova_lm
model = ols("Marks ~ Departement", data=df).fit()
anova = anova_lm(model)
print(anova)
```

```python
from statsmodels.tsa.arima.model import ARIMA # Time series analysis
# Use a defined numeric series for ARIMA.
# Here we use the cleaned Marks column.
series = clean_df["Marks"]
model = ARIMA(series, order=(1, 1, 1))
results = model.fit()
print(results.summary())
```
