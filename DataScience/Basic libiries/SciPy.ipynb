## SciPy:

```python
import scipy 
```

## importing specific model 

```python
from scipy import stats ## statistics and probability 
from scipy import optimize ## find the best solution
from scipy import linalg ## linear algebra 
```

```python
data=[10,20,30,40,50]
print(stats.tmean(data)) ## printing mean using SciPy
```

```python
data=[1,2,2,3,4]
print(stats.mode(data)) ## print mode using SciPy
```

```python
import numpy as np
data=np.array([10,20,30,40,50])
print(np.std(data))  ## printing standard Deviation using numpy 
```

## Probability Distributins 

```python
from scipy.stats import norm
data=norm.rvs(size=5)
print(data)
```

## Hypothesis Testing 

```python
from scipy.stats import ttest_1samp
data=[72,75,78,80,85]
t_stat,p_value=ttest_1samp(data,70)
print(t_stat)
print(p_value)
```

## Optimization

```python
from scipy.optimize import minimize 
def func(x):
    return x**2+5
result=minimize(func,x0=10)
print(result.x)
```

## Linear Algebra 

```python
from scipy import linalg 
```

```python
import numpy as np
A=np.array([[1,2],[3,4]])
```

```python
print(linalg.det(A)) ## Determinate 
```

```python
print(linalg.inv(A)) ## inverse
```

```python
values,vectors=linalg.eig(A)
print(values)
print(vectors)  ## Eigenvalues
```

## Numerical integration 

```python
from scipy.integrate import quad
def f(x):
    return x**2
result,error=quad(f,0,2)
print(result)
```

## Interpolation

```python
from scipy.interpolate import interp1d
x=[1,2,3,4]
y=[10,20,30,40]
f=interp1d(x,y)
print(f(2.5))
```

## Distance calculations

```python
from scipy.spatial.distance import euclidean
a=[1,2]
b=[4,6]
print(euclidean(a,b))
```
